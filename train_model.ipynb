### TALENTGUARD - FEATURE ENGINEERING

```python
#Income per Job Level
data["IncomePerJobLevel"] = (
    data["Monthly_Income"] /
    data["Job_Level"].replace(0, np.nan)
)
#Experience Ratio
data["ExperienceRatio"] = (
    data["Years_in_Current_Role"] /
    data["Years_at_Company"].replace(0, np.nan)
)
#4. Promotion Delay Flag
data["Promotion_Delay_Flag"] = (
    data["Years_Since_Last_Promotion"] >= 4
).astype(int)

# ============================================================
# CHECK THE NEW FEATURES
# ============================================================

derived_columns = ["IncomePerJobLevel","ExperienceRatio","Promotion_Delay_Flag"]
print("Derived Features Created:")
print(data[derived_columns].head())

print("\nNew Dataset Shape:")
print(data.shape)
```

### Outlier detection and handling

```python
data.columns = data.columns.str.strip()
numeric_columns = data.select_dtypes(include=[np.number]).columns
for col in numeric_columns:
    plt.figure(figsize=(6, 4))
    sns.boxplot(x=data[col])
    plt.title(f"Boxplot for {col}")
    plt.show()
```

```python
# ============================================
# OUTLIER HANDLING USING IQR CAPPING
# ============================================

# Select numerical columns
num_cols = data.select_dtypes(include=[np.number]).columns

# Create a copy for outlier handling
data_no_outliers_clip = data.copy()

# Apply IQR capping
for col in num_cols:

    Q1 = data_no_outliers_clip[col].quantile(0.25)
    Q3 = data_no_outliers_clip[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    data_no_outliers_clip[col] = data_no_outliers_clip[col].clip(
        lower=lower_bound,
        upper=upper_bound
    )


# ============================================
# BEFORE vs AFTER BOX PLOTS
# ============================================

for col in num_cols:

    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    # Before outlier handling
    sns.boxplot(y=data[col], ax=axes[0])
    axes[0].set_title(f"Before Handling: {col}")
    axes[0].set_ylabel(col)

    # After outlier handling
    sns.boxplot(y=data_no_outliers_clip[col], ax=axes[1])
    axes[1].set_title(f"After Handling: {col}")
    axes[1].set_ylabel(col)

    plt.tight_layout()
    plt.show()
```

#### Hyperparameter Tuning

```python
classifiers = {
    'Random Forest Classifier': RandomForestClassifier(random_state=42),
    'Logistic Regression': LogisticRegression(random_state=42),
    'SVM': SVC(),
    'KNN': KNeighborsClassifier(),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    
}
```

```python
param_grids = {
    'Random Forest Classifier': {
        'n_estimators':[50,100,200],
'max_depth':[5,10,15,None],
'min_samples_split':[2,5,10]
    },
    'Logistic Regression': {
        'C': [0.1, 1, 10]
    },
    'SVM': {
        'C': [1, 10],
        'kernel': ['rbf']
    },
    'KNN': {
        'n_neighbors': [3, 5]
    },
    
    'Decision Tree': {
       'max_depth':[3,5,10,15,None],
'min_samples_split':[2,5,10],
'min_samples_leaf':[1,2,3],
'criterion':['gini','entropy']
}
    }
```

```python
from sklearn.model_selection import RandomizedSearchCV
import random

n_iter_search = 20

best_models = {}
results = []

for name, clf in classifiers.items():
    print(f"\n🔍 Hyperparameter tuning for {name} using RandomizedSearchCV...")

    param_grid = param_grids.get(name, {})

    if param_grid:
        random_search = RandomizedSearchCV(
            estimator=clf,
            param_distributions=param_grid,
            n_iter=n_iter_search,
            cv=5,
            n_jobs=-1,
            verbose=1,
            random_state=42
        )

        random_search.fit(X_train_final,y_train_final)

        best_models[name] = random_search.best_estimator_

        print(f"✅ Best Parameters for {name}: {random_search.best_params_}")
        train_accuracy = accuracy_score(y_train, random_search.best_estimator_.predict(X_train))
        print(f"🎯 Training Accuracy for {name}: {train_accuracy * 100:.2f}%")

        y_pred = random_search.best_estimator_.predict(X_test)
        hyper_accuracy = accuracy_score(y_test_final, y_pred)
        print(f"🎯 Test Accuracy for {name}: {hyper_accuracy * 100:.2f}%\n")

        results.append({
            'Model': name,
            'Best Parameters': random_search.best_params_,
            'Test Accuracy': hyper_accuracy * 100,
            'Training Accuracy': train_accuracy * 100,
        })

hyper_modeling = pd.DataFrame(results)

print("\n📊 Model Performance After Randomized Search:")
hyper_modeling
```

### find the best model

```python
best_model_row = hyper_modeling.loc[hyper_modeling['Test Accuracy'].idxmax()]
best_model_name = best_model_row['Model']
best_model = best_models[best_model_name]
print(f"\n✅ Best Model Based on Test Accuracy: {best_model_name}")
print(f"📊 Best Test Accuracy: {best_model_row['Test Accuracy']:.2f}%")
```

```python
import shap
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Feature names
feature_names = X.columns

# Use your already-trained Random Forest
print("Using existing Random Forest model...")

# Create SHAP explainer
explainer = shap.TreeExplainer(rf_model)

# Convert test data to DataFrame
X_test_shap = pd.DataFrame(
    X_test_final,
    columns=feature_names
)

# Use only 50 employees for faster SHAP calculation
X_test_sample = X_test_shap.sample(
    n=min(50, len(X_test_shap)),
    random_state=42
)

print("SHAP sample size:", len(X_test_sample))

# Calculate SHAP values
print("Calculating SHAP values...")

shap_values = explainer.shap_values(
    X_test_sample
)

# Select Attrition = 1 (Leave)
if isinstance(shap_values, list):

    shap_values_leave = shap_values[1]
    base_value = explainer.expected_value[1]

elif len(shap_values.shape) == 3:

    shap_values_leave = shap_values[:, :, 1]
    base_value = explainer.expected_value[1]

else:

    shap_values_leave = shap_values
    base_value = explainer.expected_value

print("SHAP calculation completed!")


# ============================================================
# SHAP SUMMARY PLOT
# ============================================================

shap.summary_plot(
    shap_values_leave,
    X_test_sample,
    max_display=10
)


# ============================================================
# SHAP FEATURE IMPORTANCE
# ============================================================

shap.summary_plot(
    shap_values_leave,
    X_test_sample,
    plot_type="bar",
    max_display=10
)


# ============================================================
# SHAP IMPORTANCE TABLE
# ============================================================

shap_importance = pd.DataFrame({
    "Feature": feature_names,
    "Mean SHAP Value":
        np.abs(shap_values_leave).mean(axis=0)
})

shap_importance = shap_importance.sort_values(
    by="Mean SHAP Value",
    ascending=False
).reset_index(drop=True)

print("\nTop 10 Attrition Drivers:")

display(
    shap_importance.head(10)
)


# ============================================================
# WATERFALL PLOT
# ============================================================

sample_index = 0

shap.waterfall_plot(
    shap.Explanation(
        values=shap_values_leave[sample_index],
        base_values=base_value,
        data=X_test_sample.iloc[sample_index],
        feature_names=feature_names
    )
)
```
